```python
# data_loader.py: Data Loader Utilities
from pathlib import Path
import json
from datasets import load_dataset
from huggingface_hub import login

# Create target path
current_dir = Path.cwd()
project_root = current_dir.parent
raw_path = project_root / 'data' / 'raw'
metadata_path = project_root / 'data' / 'metadata'

# Create directories if they don't exist
raw_path.mkdir(parents=True, exist_ok=True)
metadata_path.mkdir(parents=True, exist_ok=True)
```

```python
# Validate paths
print(project_root)
print(raw_path)
print(metadata_path)
```

```python
# Load Hugging Face dataset (Natural Disasters) in data/raw
ds = load_dataset(
    "community-datasets/disaster_response_messages",
    )

# test, train, split - melisekm/natural-disasters-from-social-media
# test1, train1, validation1 - community-datasets/disaster_response_messages
```

```python
# Save dataset to my folder
for split in ["train", "test", "validation"]:
    file_path = raw_path / f"{split}1.csv"
    ds[split].to_csv(file_path, index=False)
    print(f"Saved {split} to: {file_path}")  # Verification
```

```python
# Save metadata
metadata = {
    "dataset_name": "natural-disasters-from-social-media",
    "description": ds['train'].info.description,
    "features": str(ds["train"].features),
    "num_samples": {split: len(ds[split]) for split in ds.keys()}
}

with open(metadata_path / "dataset_metadata.json", "w") as f:
    json.dump(metadata, f, indent=2)
```
